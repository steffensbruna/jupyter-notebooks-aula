### Questão 1: Orçamento de viagem (1,0 ponto)

Escreva um programa que calcule o custo de uma viagem de carro.

O programa deve ler do usuário, usando `input`:

- a distância da viagem, em km (`float`);
- o consumo médio do carro, em km/litro (`float`);
- o preço do litro do combustível, em reais (`float`);
- o valor total dos pedágios, em reais (`float`);
- o número de pessoas que dividirão a despesa (`int`).

Em seguida o programa deve calcular e exibir:

1. a quantidade de litros necessária (distância dividida pelo consumo);
2. o custo do combustível;
3. o custo total (combustível + pedágios);
4. o custo por pessoa;
5. o tipo (`type`) das variáveis da distância e do número de pessoas.

Todos os valores em reais devem ser exibidos com **2 casas decimais** (use `round` ou formatação de string).

Exemplo:

```
Distância da viagem (km): 450
Consumo médio (km/l): 12.5
Preço do litro (R$): 5.89
Total de pedágios (R$): 32.40
Número de pessoas: 3

Litros necessários: 36.0
Custo do combustível: R$ 212.04
Custo total da viagem: R$ 244.44
Custo por pessoa: R$ 81.48
Tipo da distância: <class 'float'> | Tipo do número de pessoas: <class 'int'>
```

````python
distance = float(input("Distância da Viagem (km): "))
consume = float(input("Consumo Médio (km/l): "))
fuel_price = float(input("Preço do Litro (R$): "))
toll = float(input("Total de Pedágios (R$): "))
people = int(input("Número de Pessoas: "))

#OPERAÇÕES
liters = distance/consume
total_fuel_price = fuel_price * liters
total_price = total_fuel_price + toll

print("")
print("")
print("===========================================")
print("           ORÇAMENTO DE VIAGEM             ")
print("===========================================")
print(f"Litros necessários: {liters:.1f}")
print(f"Custo do Combustível: R$ {total_fuel_price:.2f}")
print(f"Custo Total da Viagem: R$ {total_price:.2f}")
print(f"Custo por Pessoa R$ {total_price/people:.2f}")
print(f"Tipo da distância: {type(distance)} | Tipo do número de pessoas: {type(people)}")
````

### Questão 3: Distância entre dois pontos (1,0 ponto)

Usando a instrução `import` e o módulo `math`, escreva um programa que:

1. leia as coordenadas de dois pontos do plano: `x1`, `y1`, `x2` e `y2` (todos `float`);
2. calcule `dx` e `dy` como as diferenças **em valor absoluto** entre as coordenadas;
3. calcule a distância entre os pontos com a fórmula abaixo:

$$distancia = \sqrt{dx^2 + dy^2}$$

4. calcule a área do círculo que tem essa distância como raio;
5. exiba `dx`, `dy`, a distância e a área, todos arredondados para 2 casas decimais com `round`.

Exemplo:

```
x1: 1
y1: 2
x2: 4
y2: 6

dx = 3.0
dy = 4.0
Distância entre os pontos: 5.0
Área do círculo de raio 5.0: 78.54
```

````python
import math

x1 = float(input("x1: "))
y1 = float(input("y1: "))
x2 = float(input("x2: "))
y2 = float(input("y2: "))

#OPERAÇÕES
dx = abs(x1-x2)
dy = abs(y1-y2)
distance = math.sqrt(math.pow(dx, 2) + math.pow(dy, 2))
area = math.pi * math.pow(distance, 2) # pi * raio²


print("")
print(f"dx = {dx}")
print(f"dy = {dy}")
print(f"Distância entre os pontos: {distance}")
print(f"Área do Círculo de raio {distance}: {round(area, 2)}")
````

### Questão 4: Encapsulamento e refatoração (1,0 ponto)

O código abaixo calcula e exibe a média de três alunos. Ele **funciona**, mas repete a mesma lógica três vezes:

```python
nota1 = 7.0
nota2 = 5.5
nota3 = 9.0
media = (nota1 + nota2 + nota3) / 3
print("Média de Ana:", round(media, 2))

nota1 = 4.0
nota2 = 6.5
nota3 = 5.0
media = (nota1 + nota2 + nota3) / 3
print("Média de Bruno:", round(media, 2))

nota1 = 9.5
nota2 = 8.0
nota3 = 10.0
media = (nota1 + nota2 + nota3) / 3
print("Média de Carla:", round(media, 2))
```

Sua tarefa:

**(a)** **Encapsule** essa lógica em uma função chamada `exibir_media(nome, nota1, nota2, nota3)`, que calcula a
média das três notas e exibe a mensagem no mesmo formato. A função deve ter uma **docstring** descrevendo o que
ela faz e o que significa cada parâmetro.

**(b)** Chame a função três vezes para reproduzir exatamente a mesma saída do código original.

**(c)** Na célula de texto indicada, explique com suas palavras: o que é **encapsulamento**, o que é
**refatoração**, e por que a versão com função é melhor que a original.

Saída esperada:

```
Média de Ana: 7.17
Média de Bruno: 5.17
Média de Carla: 9.17
```


````python
def exibir_media(nome, nota1, nota2, nota3):
  '''MÉDIA DOS ALUNOS: função criada para otimização
    do código, onde realiza o cálculo de média das notas.
    1) soma as 3 notas e divide por 3
    2) mostra na tela o nome do(a) aluno(a) e a média arredondada
    com 2 casas decimais.
  '''
  media = (nota1 + nota2 + nota3) / 3 #1
  print(f"Média de {nome}: {round(media,2)}") #2

exibir_media("Ana", 7.0, 5.5, 9.0)
exibir_media("Bruno", 4.0, 6.5, 5.0)
exibir_media("Carla", 9.5, 8.0, 10.0)
````

**(c) Sua explicação sobre encapsulamento e refatoração:**

Refatoração é melhorar a estrutura de um código. Uma ótima forma de fazer isso é através do encapsulamento, que consiste na criação de métodos que manipulam dados, podendo ser reutilizado várias e várias vezes.

### Questão 5: Generalização, parâmetros e variáveis locais (1,0 ponto)

**(a)** Escreva uma função chamada `moldura(texto, caractere, largura)` que exibe um texto dentro de uma moldura.
A função deve:

- imprimir uma linha formada pela repetição de `caractere`, com a `largura` especificada;
- imprimir o `texto` entre dois `caractere` (um no início e outro no fim da linha);
- imprimir novamente a linha de fechamento.

A função deve ter uma **docstring**. Note que ela **não retorna valor**, apenas exibe (retorna `None`).

Exemplo:

```
moldura("Trabalho 1", "*", 20)
```

Saída:

```
********************
* Trabalho 1
********************
```

**(b)** Chame `moldura` **três vezes**, com textos, caracteres e larguras diferentes, mostrando que a função foi
**generalizada** pelos parâmetros.

````python
def moldura(texto, caractere, largura):
  '''MOLDURA: faz uma moldura totalmente visual de forma
  a incrementar o visual da saída.
  1) printa o caractere pela quantidade de vezes da largura
  2) print o texto informado entre os caracteres
  3) novamente printa o caractere pela quantidade de vezes da largura
  '''
  print(caractere * largura)
  print(caractere + " " + texto + " " + caractere)
  print(caractere * largura)

moldura("Trabalho 1", "*", 20)
print("")
moldura("PYTHON", "#", 15)
print("")
moldura(" ARTE ", "=", 10)
````
