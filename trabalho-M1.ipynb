### Questão 1: Orçamento de viagem (1,0 ponto)

Escreva um programa que calcule o custo de uma viagem de carro.

- o consumo médio do carro, em km/litro (`float`);
- o preço do litro do combustível, em reais (`float`);
- o valor total dos pedágios, em reais (`float`);
- o número de pessoas que dividirão a despesa (`int`).

Em seguida o programa deve calcular e exibir:

1. a quantidade de litros necessária (distância dividida pelo consumo);
2. o custo do combustível;
3. o custo total (combustível + pedágios);

### Questão 3: Distância entre dois pontos (1,0 ponto)

Usando a instrução `import` e o módulo `math`, escreva um programa que:

1. leia as coordenadas de dois pontos do plano: `x1`, `y1`, `x2` e `y2` (todos `float`);
2. calcule `dx` e `dy` como as diferenças **em valor absoluto** entre as coordenadas;
3. calcule a distância entre os pontos com a fórmula abaixo:

$$distancia = \sqrt{dx^2 + dy^2}$$

4. calcule a área do círculo que tem essa distância como raio;
5. exiba `dx`, `dy`, a distância e a área, todos arredondados para 2 casas decimais com `round`.

Exemplo:

```
x1: 1
y1: 2
x2: 4
y2: 6

dx = 3.0
dy = 4.0
Distância entre os pontos: 5.0
Área do círculo de raio 5.0: 78.54
```

````python
import math

x1 = float(input("x1: "))
y1 = float(input("y1: "))
x2 = float(input("x2: "))
y2 = float(input("y2: "))

#OPERAÇÕES
dx = abs(x1-x2)
dy = abs(y1-y2)
distance = math.sqrt(math.pow(dx, 2) + math.pow(dy, 2))
area = math.pi * math.pow(distance, 2) # pi * raio²


print("")
print(f"dx = {dx}")
print(f"dy = {dy}")
print(f"Distância entre os pontos: {distance}")
print(f"Área do Círculo de raio {distance}: {round(area, 2)}")
````

### Questão 4: Encapsulamento e refatoração (1,0 ponto)

O código abaixo calcula e exibe a média de três alunos. Ele **funciona**, mas repete a mesma lógica três vezes:

```python
nota1 = 7.0
nota2 = 5.5

media = (nota1 + nota2 + nota3) / 3
print("Média de Ana:", round(media, 2))

nota1 = 9.5

media = (nota1 + nota2 + nota3) / 3
print("Média de Carla:", round(media, 2))
```

Sua tarefa:

**(a)** **Encapsule** essa lógica em uma função chamada `exibir_media(nome, nota1, nota2, nota3)`, que calcula a
média das três notas e exibe a mensagem no mesmo formato. A função deve ter uma **docstring** descrevendo o que
ela faz e o que significa cada parâmetro.

**(b)** Chame a função três vezes para reproduzir exatamente a mesma saída do código original.

**(c)** Na célula de texto indicada, explique com suas palavras: o que é **encapsulamento**, o que é
**refatoração**, e por que a versão com função é melhor que a original.

Saída esperada:

```
Média de Ana: 7.17
Média de Bruno: 5.17
Média de Carla: 9.17
```


````python
def exibir_media(nome, nota1, nota2, nota3):
  '''MÉDIA DOS ALUNOS: função criada para otimização
    do código, onde realiza o cálculo de média das notas.
    1) soma as 3 notas e divide pela quantidade de notas
    2) mostra na tela o nome do(a) aluno(a) e a média arredondada
    com 2 casas decimais.
  '''
  media = (nota1 + nota2 + nota3) / 3 #1
  print(f"Média de {nome}: {round(media,2)}") #2

exibir_media("Ana", 7.0, 5.5, 9.0)
exibir_media("Bruno", 4.0, 6.5, 5.0)
exibir_media("Carla", 9.5, 8.0, 10.0)
````

**(c) Sua explicação sobre encapsulamento e refatoração:**

Refatoração é melhorar a estrutura de um código. Uma ótima forma de fazer isso é através do encapsulamento, que consiste na criação de métodos que manipulam dados, podendo ser reutilizado várias e várias vezes. A versão com função é infinitamente melhor pois a operação de média não precisa ser feita 3 vezes, basta chamar a função que ela já retorna o resultado da operação. Isso economiza tempo e espaço na memória.

Exemplo:

```

```

````python
from os.path import basename, exists

def download(url):
    filename = basename(url)
    if not exists(filename):
        from urllib.request import urlretrieve

        local, _ = urlretrieve(url, filename)
        print("Downloaded " + str(local))
    return filename

download('https://github.com/ramalho/jupyturtle/releases/download/2024-03/jupyturtle.py');
````

````python
# (a) e (b) defina aqui as funções poligono e flor

from jupyturtle import forward, make_turtle, left, right

def poligono(lados, tamanho):
  '''
  -> desenha um polígono a partir dos dados informados de lados e tamanho.
  VARIÁVEIS:
  lados: quantos lados terá o polígono, e valor usado para descobrir o ângulo
  tamanho: valor usado para o tamanho que o jupyturtle deve percorrer

  '''
  for i in range (lados):
    forward(tamanho)
    left(360/lados)

def flor(petalas,lados,tamanho):
  '''
  -> desenha uma flor usando polígonos a partir da função poligono (que desenha
  um polígono)
  VARIAVEIS:
  petalas: quantidade de pétalas (quantidade de polígonos à ser desenhado)
  lados: valor utilizado para obter o ângulo
  tamanho: valor usado para o tamanho que a jupyturtle deve percorrer

  '''
  for i in range (petalas):
      poligono(lados, tamanho)
      left(360/lados)
````

````python
# (c) triângulo

make_turtle()
poligono(3,50)
````

````python
# (c) hexágono

make_turtle()
poligono(6,30)
````

````python
# (c) flor

make_turtle()
flor(7, 7, 30)
````

### Questão 8: Condicionais encadeadas e aninhadas (1,0 ponto)

Escreva uma função chamada `classificar_imc(peso, altura)` que:

1. **valide os dados**: se `peso` **ou** `altura` for menor ou igual a zero, a função deve
   retornar `None`, sem calcular nada;
2. caso os dados sejam válidos, calcule o IMC com a fórmula `imc = peso / altura ** 2`;
3. use uma **condicional encadeada** (`if`/`elif`/`else`) para **retornar** a classificação correspondente:

| IMC | Classificação |
|---|---|
| menor que 18,5 | `"Abaixo do peso"` |
| de 18,5 a 24,9 | `"Peso normal"` |
| de 25,0 a 29,9 | `"Sobrepeso"` |
| 30,0 ou mais | `"Obesidade"` |

Depois, escreva um programa que leia o peso e a altura do usuário, chame a função e:

- se o retorno for `None`, exiba a mensagem `"Dados inválidos: peso e altura devem ser positivos."`;
- caso contrário, exiba o IMC arredondado com 2 casas decimais e a classificação.

A função deve ter **docstring**.

Exemplo:

```
Peso (kg): 72.5
Altura (m): 1.75
IMC: 23.67 -> Peso normal
```

```
Peso (kg): 70
Altura (m): 0
Dados inválidos: peso e altura devem ser positivos.
```

````python
def classificar_imc(peso, altura):
    '''
    Realiza a validação dos dados informados, caso seja validado,
    é feito o cálculo do IMC e armazenado na variável "imc" e retornado a
    classificação. Caso os dados não sejam válidos, retorna "None".

    imc = cálculo do imc
    peso = float
    altura = float

    '''
    if peso <= 0 or altura <= 0:
      return
    imc = peso / altura ** 2
    if imc < 18.5:
      return "Abaixo do peso"
    elif imc <= 24.9:
      return "Peso normal"
    elif imc <= 29.9:
      return "Sobrepeso"
    else:
      return "Obesidade"

peso = (float(input("Peso (kg): ")))
altura = (float(input("Altura (m): ")))

if classificar_imc(peso,altura) == None:
  print("Dados inválidos: peso e altura devem ser positivos.")
else:
  print(f"IMC: {round(peso / altura ** 2,2)} -> {classificar_imc(peso,altura)}")
````

````python
# Versão 1 - somente as funções, testadas com valores fixos

def media_notas(nota1, nota2, nota3):
    '''
    realiza a soma das notas e divide pela quantidade
    de notas para obter a média do(a) aluno(a).

    '''
    media = (nota1+nota2+nota3)/3
    return media

def situacao(media):
    '''
    a partir do retorno da função media_notas, realiza
    a classificação do(a) aluno(a).
    '''
    if media >= 7:
        return "Aprovado"
    elif media >= 4:
        return "Recuperação"
    else:
        return "Reprovado"

print(round(media_notas(8.0,7.6,9.5),2))
situacao(media_notas(8.0,7.6,9.5))
````

````python
# Versão 2 - laço de leitura e linhas do boletim

quantidade = int(input("Quantidade de Alunos(a)? "))
boletim = ""

for i in range (quantidade):
    nome = input(f"Aluno {i+1} - nome: ")
    nota1 = float(input("Nota 1: "))
    nota2 = float(input("Nota 2: "))
    nota3 = float(input("Nota 3: "))
    media = media_notas(nota1,nota2,nota3)
    status = situacao(media)
    boletim = boletim + f"{nome}: {media:.1f} - {status}\n"

print("")
print("=========================")
print("         BOLETIM         ")
print("=========================")
print(boletim, end="")
print("=========================")
````

````python
# Versão 3 - versão final, com as estatísticas da turma

quantidade = int(input("Quantidade de Alunos(a)? "))
boletim = ""
total_notas: float = 0
aprovados = 0
melhorMedia = 0
melhorAluno = ""

for i in range (quantidade):
    nome = input(f"Aluno {i+1} - nome: ")
    nota1 = float(input("Nota 1: "))
    nota2 = float(input("Nota 2: "))
    nota3 = float(input("Nota 3: "))
    media = media_notas(nota1,nota2,nota3)
    status = situacao(media)
    boletim = boletim + f"{nome}: {media:.1f} - {status}\n"
    total_notas += media

    if media > melhorMedia:
        melhorAluno = nome
        melhorMedia = media

    if status == "Aprovado":
        aprovados += 1


print("")
print("=========================")
print("         BOLETIM         ")
print("=========================")
print(boletim, end="")
print("=========================")
print(f"Média geral da turma: {total_notas/quantidade:.2f}")
print(f"Aprovados: {aprovados} de {quantidade} ({(aprovados/quantidade)*100:.2f})")
print(f"Melhor média: {melhorAluno} ({melhorMedia:.2f})")
````
